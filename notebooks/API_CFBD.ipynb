```python
import os
import json
import urllib.request
import urllib.parse
import pandas as pd
```

```python
# Load API key from environment variable for securityimport osAPI_KEY = os.environ.get('CFBD_API_KEY', 'YOUR_API_KEY_HERE')headers = {    "Authorization": f"Bearer {API_KEY}"}def cfbd_get(url, params=None):    if params:        url = url + "?" + urllib.parse.urlencode(params)        req = urllib.request.Request(url, headers=headers)        with urllib.request.urlopen(req) as response:        data = response.read().decode()        return json.loads(data)all_games = []for year in range(2002,2024+1):    print(f"Fetchin {year}...")    data = cfbd_get("https://api.collegefootballdata.com/games", {"year": year})    all_games.extend(data)games_df = pd.DataFrame(all_games)print("Games sample:")display(games_df.head())venues = cfbd_get("https://api.collegefootballdata.com/venues")venues_df = pd.DataFrame(venues)print("Venues sample:")display(venues_df.head())merged = games_df.merge(    venues_df,    left_on="venue",    right_on="name",    how="left")print("Merged sample:")display(merged.head())
```

```python
output_path = "../data/raw/cfbd_merged.csv"

merged.to_csv(output_path, index=False)

print(f"CSV exported to: {output_path}")
```
